## USB command protocol notes

This notebook is intentionally limited to USB serial communication and raw data capture. It does **not** decode spectra. Offline decoding experiments live in `06_scio_offline_decode.ipynb`.

The Android app frames SCiO protocol commands as:

`[sequence, 0xBA, command, payload_length_low, payload_length_high, payload...]`

For a first packet, `sequence = 1`. The command byte is signed in the Java decompilation, but serial bytes are written as unsigned values from `0x00` to `0xFF`. Multi-byte payload lengths and integer payloads observed in the app are little-endian.

| Command | Signed byte | Hex | Payload | Response | App evidence | Risk |
|---|---:|---:|---|---|---|---|
| READ_DEVICE_STATUS | 0 | 0x00 | none | status payload | `CommandIDs.READ_DEVICE_STATUS` | read-only |
| READ_DEVICE_ID | 1 | 0x01 | none | DSP/Aptina identifiers | `performReadDeviceId` | read-only |
| SAMPLE_SPECTRUM | 2 | 0x02 | none | dark, sample, optional gradient blobs | `performSpectrum` / scan flow | capture |
| READ_TEMPERATURE | 4 | 0x04 | none | temperature words | `performReadTemperature` | read-only |
| READ_BATTERY_STATE | 5 | 0x05 | none | battery fields | `performReadBattery` | read-only |
| READ_EVENT_LOG | 6 | 0x06 | none | unknown/log payload | `CommandIDs.READ_EVENT_LOG` | read-only experimental |
| PARAMETER_SET | 7 | 0x07 | parameter/value payload | ACK | `CommandIDs.PARAMETER_SET` | write, guarded |
| PARAMETER_GET | 8 | 0x08 | parameter id | parameter payload | `CommandIDs.PARAMETER_GET` | read-only experimental |
| SET_INDICATION_LED | 11 | 0x0B | LED payload | ACK | existing notebook/app | write, guarded |
| READY_FOR_WR | 14 | 0x0E | none | ACK | `performReadyForWR` | write/state, guarded |
| CLEAR_READY_FOR_WR | 17 | 0x11 | none | ACK | `performClearReadyForWR` | write/state, guarded |
| READ_FILE_HEADER | -121 | 0x87 | `<I file_id` | four `u32`; checksum is word 4 | `performReadFileHeader` | read-only |
| READ_FILE_LIST | -108 | 0x94 | none | 8-byte file entries | `performReadFileList` | read-only |
| WRITE_USER_DEVICE_NAME | -111 | 0x91 | name payload | ACK | `CommandIDs.WRITE_USER_DEVICE_NAME` | write, guarded |
| READ_BLE_ID | -124 | 0x84 | none | BLE id/name/I2S tag | `performReadDeviceBleId` | read-only |
| READ_BLE_STATUS | -123 | 0x85 | none | BLE status | `CommandIDs.READ_BLE_STATUS` | read-only |
| RESET_DEVICE | -125 | 0x83 | none | reset/disconnect | `CommandIDs.RESET_DEVICE` | disruptive, guarded |
| READ_BLE | -101 | 0x9B | none | BLE config | `CommandIDs.READ_BLE` | read-only |
| WRITE_BLE | -102 | 0x9A | BLE value payload | ACK | `CommandIDs.WRITE_BLE` | write, guarded |
| FILE_DOWNLOAD | -127 | 0x81 | firmware frame payload | ACK | `performFileDownload` | host-to-device write, guarded |

Firmware/image-to-spectrum parameter files referenced by the Android app are:

| Name | File ID | Notes |
|---|---:|---|
| `deadPixelsIndices` | 100 | device/I2S-specific correction data |
| `centers` | 101 | likely bin/stripe center data |
| `bins` | 102 | likely image-to-spectrum bin definitions |
| `nPixelsPerBin` | 103 | likely per-bin normalization/count data |


```python
import json
import struct
import logging
import datetime
import os
import csv
import base64
import serial
import serial.tools.list_ports as pyserial
import numpy as np

# Logging/output format setup
log = logging.getLogger('root')
log.setLevel(logging.DEBUG)
logging.basicConfig(format='[%(asctime)s] %(levelname)8s: %(message)s', datefmt='%Y-%m-%d %H:%M:%S')
```

```python
#### Search for device
def search_scio_usb(silent=False):
    if not silent: print('Searching for devices:')
    scio_ID = '0451:16AA'
    device_list = []
    # Find the scio port. Returns an empty list if not found
    scio_port = list(pyserial.grep(scio_ID))
    if(len(scio_port) == 0):
        print('  - Device not found. Try a long press (device should respond with blue blinking)')
        return(device_list)
    device_list.append({'address': scio_port[0].device, 'id': scio_ID})

    return(device_list)

class scio_usb:
    def __init__(self, address):
        self.address = address
        self.client = None
        self.device_info = {}
        self.scan_rawdata = {}
        self.scan_b64data = {}
        self.temperature = {}
        self.calibration_data = {}
        self.key = b''
        self.disconnected = True
        # Android CommandIDs.java constants. Java shows signed bytes; serial writes use unsigned byte values.
        self.cmd_base_protocol =  -70 # 0xBA, protocol marker in every first command packet
        self.cmd_dev_status    =    0 # 0x00, read-only
        self.cmd_dev_id        =    1 # 0x01, read-only DSP/Aptina identifiers
        self.cmd_scan          =    2 # 0x02, capture dark/sample/gradient raw blobs
        self.cmd_temp          =    4 # 0x04, read-only temperature words
        self.cmd_bat           =    5 # 0x05, read-only battery state
        self.cmd_event_log     =    6 # 0x06, read-only/experimental event log
        self.cmd_param_set     =    7 # 0x07, write; guarded by allow_write
        self.cmd_param_get     =    8 # 0x08, read-only/experimental parameter read
        self.cmd_bist          =    9 # 0x09, built-in self-test; experimental
        self.cmd_led           =   11 # 0x0B, write; guarded by allow_write in generic helpers
        self.ready_for_wr      =   14 # 0x0E, write/state; prepare for white reference
        self.clear_ready_wr    =   17 # 0x11, write/state; clear white-reference readiness
        self.cmd_file_dload    = -127 # 0x81, host-to-device firmware frame write; guarded
        self.cmd_reset_dev     = -125 # 0x83, disruptive reset; guarded
        self.cmd_ble_id        = -124 # 0x84, read-only BLE id/name/I2S tag
        self.cmd_ble_status    = -123 # 0x85, read-only BLE status
        self.cmd_file_head     = -121 # 0x87, read-only file header/checksum; payload is <I file_id
        self.cmd_set_name      = -111 # 0x91, write user-visible BLE name; guarded
        self.cmd_file_list     = -108 # 0x94, read-only file list
        self.cmd_write_ble     = -102 # 0x9A, write BLE setting; guarded
        self.cmd_get_ble       = -101 # 0x9B, read-only BLE config/status
        self.firmware_param_files = {
            'deadPixelsIndices': 100,
            'centers': 101,
            'bins': 102,
            'nPixelsPerBin': 103,
        }
        
    async def connect(self):
        try:
            self.client = serial.Serial(self.address)
            self.disconnected = False
        except Exception as e:
            logging.error(f'Failed to connect to {self.address}: {e}')
            raise
        pass
    
    async def disconnect(self):
        if not self.disconnected:
            try:
                self.client.close()
            except Exception as e:
                logging.error(f'Failed to disconnect from {self.address}: {e}')
                raise
            finally:
                self.disconnected = True
        pass
        
    @staticmethod
    def _to_u8(value):
        # Java bytes in the decompiled app are signed. The serial wire still carries 0..255 bytes.
        return value & 0xFF

    def create_command(self, cmd, payload=b'', seq=1):
        """Build one SCiO protocol command packet.

        Packet shape from RequestCommandBuilder.java:
            [sequence, 0xBA, command, payload_length_low, payload_length_high, payload...]

        Most read commands have an empty payload. Payload commands such as READ_FILE_HEADER
        pass little-endian binary data after the 2-byte length.
        """
        if payload is None:
            payload = b''
        if not isinstance(payload, (bytes, bytearray)):
            raise TypeError('payload must be bytes-like')
        if len(payload) > 0xFFFF:
            raise ValueError('payload too large for 16-bit SCiO command length')
        header = bytes([
            self._to_u8(seq),
            self._to_u8(self.cmd_base_protocol),
            self._to_u8(cmd),
            len(payload) & 0xFF,
            (len(payload) >> 8) & 0xFF,
        ])
        return header + bytes(payload)
    
    async def send_command(self, byte_cmd):
        try:
            self.client.write(byte_cmd)
        except Exception as e:
            logging.error(f'Failed to send command to {self.address}: {e}')
            raise
        pass
    
    async def read_response(self):
        try:
            # Start reading the response
            s = self.client.read(1)
        except Exception as e:
            logging.error(f'Failed to read command from {self.address}: {e}')
            raise
        response_type = struct.unpack('<b',s)[0]
        # Error check
        assert (response_type == self.cmd_base_protocol), 'Wrong response type: {}'.format(response_type)
        # Read data
        s = self.client.read(1)
        response_content = struct.unpack('<b',s)[0]
        s = self.client.read(2)
        response_length = struct.unpack('<H',s)[0]
        # Read the number of values specified by the response length
        response_data = self.client.read(response_length)
        return([response_content, response_length, response_data])
    
    async def read_temperature(self):
        byte_cmd = self.create_command(self.cmd_temp)
        await self.send_command(byte_cmd)
        [response_content, response_length, response_data_raw] = await self.read_response()
        # Parse response
        num_vars = response_length / 4 # divide by 4 because we are dealing with longs
        data_struct = '<' + str(int(num_vars)) + 'L' # This is usually '<3l' or '<lll'
        # Convert bytes
        response_data = struct.unpack(data_struct, response_data_raw)
        # Convert to actual temperatures
        cmosTemperature   = (response_data[0] - 375.22) / 1.4092 # From the disassembled Android app
        chipTemperature   = response_data[1] / 100
        objectTemperature = response_data[2] / 100
        temperatures = {'cmos_t': cmosTemperature, 'chip_t': chipTemperature, 'obj_t': objectTemperature}
        self.temperature.update(temperatures)
        return(temperatures)
    
    async def read_file_list(self):
        """Read firmware file list from the device.

        Android evidence: SCiOBLeService.performReadFileList() sends READ_FILE_LIST (0x94).
        The SDK parses the response as 8-byte entries: little-endian u32 file_type, u32 file_version.
        This is read-only and safe to use for inventory.
        """
        byte_cmd = self.create_command(self.cmd_file_list)
        await self.send_command(byte_cmd)
        [response_content, response_length, response_data_raw] = await self.read_response()
        file_list = {}
        for offset in range(0, len(response_data_raw) - 7, 8):
            file_type, file_version = struct.unpack_from('<II', response_data_raw, offset)
            file_nb = offset // 8
            file_list[file_nb] = {
                'file_type': file_type,
                'file_version': file_version,
                'file_type_hex': response_data_raw[offset:offset+4].hex(),
                'file_version_hex': response_data_raw[offset+4:offset+8].hex(),
            }
        return file_list
    
    async def read_file_header(self, file_id):
        """Read the 16-byte header/checksum block for a firmware file id.

        Android evidence: SCiOBLeService.performReadFileHeader(int) sends READ_FILE_HEADER
        with ByteBuffer.allocate(4).order(LITTLE_ENDIAN).putInt(file_id). The app reads four
        little-endian u32 values and treats the fourth word (offset 12) as the checksum.
        """
        payload = struct.pack('<I', int(file_id))
        byte_cmd = self.create_command(self.cmd_file_head, payload=payload)
        await self.send_command(byte_cmd)
        [response_content, response_length, response_data_raw] = await self.read_response()
        words = list(struct.unpack('<4I', response_data_raw[:16])) if len(response_data_raw) >= 16 else []
        return {
            'file_id': int(file_id),
            'response_command': response_content,
            'response_length': response_length,
            'raw_hex': response_data_raw.hex(),
            'words_u32le': words,
            'checksum': words[3] if len(words) >= 4 else None,
        }

    async def read_firmware_param_headers(self):
        """Read headers/checksums for the four image-to-spectrum parameter files.

        These files are the most likely missing offline-decoder inputs:
        deadPixelsIndices(100), centers(101), bins(102), nPixelsPerBin(103).
        This method reads only headers/checksums; it does not download or write firmware data.
        """
        headers = {}
        for name, file_id in self.firmware_param_files.items():
            headers[name] = await self.read_file_header(file_id)
        return headers

    async def read_parameter(self, parameter_id):
        """Experimental read-only PARAMETER_GET command.

        The app defines PARAMETER_GET (0x08), but this notebook has not proven the exact
        payload/response schema. The safest hypothesis is a little-endian u32 parameter id.
        """
        payload = struct.pack('<I', int(parameter_id))
        byte_cmd = self.create_command(self.cmd_param_get, payload=payload)
        await self.send_command(byte_cmd)
        [response_content, response_length, response_data_raw] = await self.read_response()
        return {
            'parameter_id': int(parameter_id),
            'response_command': response_content,
            'response_length': response_length,
            'raw_hex': response_data_raw.hex(),
            'raw_bytes': response_data_raw,
        }

    def _require_write_allowed(self, allow_write):
        # Side-effectful commands stay opt-in so accidental notebook execution cannot alter the device.
        if not allow_write:
            raise RuntimeError('This command writes to or changes device state. Pass allow_write=True intentionally.')

    async def parameter_set(self, parameter_id, value, allow_write=False):
        """Guarded PARAMETER_SET helper. Payload shape is experimental: <II parameter_id,value."""
        self._require_write_allowed(allow_write)
        payload = struct.pack('<II', int(parameter_id), int(value))
        byte_cmd = self.create_command(self.cmd_param_set, payload=payload)
        await self.send_command(byte_cmd)
        return await self.read_response()

    async def file_download(self, frame_payload, frame_seq=0, allow_write=False):
        """Guarded FILE_DOWNLOAD helper.

        Android evidence indicates this is host-to-device firmware write, not device-to-host read.
        Use only when intentionally reproducing firmware upgrade writes.
        """
        self._require_write_allowed(allow_write)
        byte_cmd = self.create_command(self.cmd_file_dload, payload=bytes(frame_payload), seq=int(frame_seq) + 1)
        await self.send_command(byte_cmd)
        return await self.read_response()

    async def ready_for_wr(self, allow_write=False):
        """Guarded READY_FOR_WR command; changes white-reference state."""
        self._require_write_allowed(allow_write)
        await self.send_command(self.create_command(self.ready_for_wr))
        return await self.read_response()

    async def clear_ready_for_wr(self, allow_write=False):
        """Guarded CLEAR_READY_FOR_WR command; changes white-reference state."""
        self._require_write_allowed(allow_write)
        await self.send_command(self.create_command(self.clear_ready_wr))
        return await self.read_response()

    async def reset_device(self, allow_write=False):
        """Guarded RESET_DEVICE command; may disconnect/reboot the SCiO."""
        self._require_write_allowed(allow_write)
        await self.send_command(self.create_command(self.cmd_reset_dev))
        return await self.read_response()
    
    async def read_battery(self):
        # Firmware-version dependent handling (unclear if this is needed)
        # Check if firmware version was already read, to figure out if firmware version is available
        #if(len(self.device_info) == 0):
        #    await self.read_device_id()
        #if(self.device_info['firmwareVersion'] >= 149):
        #    #get_percent
        #    pass
        #else:
        #    #get_volt
        #    pass
            
        # Now read the information
        byte_cmd = self.create_command(self.cmd_bat)
        await self.send_command(byte_cmd)
        [response_content, response_length, response_data_raw] = await self.read_response()
        
        # Parse
        data_struct = '<HBBHH8x' # Battery logic
        resp = list(struct.unpack(data_struct, response_data_raw))
        # Battery voltage
        resp[4] = resp[4]/1000
        # Charge status
        resp[3] = resp[3] & 3
        if(resp[3] == 0):
            charge_status = 'not charging'
        elif(resp[3] == 4):
            charge_status = 'full'
        elif(resp[3] != 6):
            charge_status = 'charging'
        else:
            charge_status = 'battery error'
        # Store information
        battery_status = {'bat_charge_percent': resp[0],
                          'bat_health_percent': resp[1],
                          'bat_health_status':  resp[2],
                          'bat_charge_status':  charge_status,
                          'bat_voltage':        resp[4]}
        return(battery_status)
    
    async def read_ble_id(self):
        byte_cmd = self.create_command(self.cmd_ble_id)
        await self.send_command(byte_cmd)
        [response_content, response_length, response_data_raw] = await self.read_response()
        # Parse response
        data_struct = '<8xH40s16s64s'
        ble_data = list(struct.unpack(data_struct, response_data_raw))
        ble_data[1] = ble_data[1].decode('utf-8').strip()       # Serial number
        ble_data[2] = ble_data[2].decode('utf-8').strip('\x00') # Device name
        ble_data[3] = ble_data[3].decode('utf-8').strip()       # Production information
        # Store information
        # Note: The remaining (first) 8 bytes are a device BLE ID in Hex
        self.device_info.update({'deviceBleId':    response_data_raw[0:8].hex(),
                                 'bleFWVersion':   ble_data[0],
                                 'serial_number':  ble_data[1],
                                 'device_name':    ble_data[2],
                                 'i2s_tag_config': ble_data[3]})
        return(self.device_info)
    
    async def read_device_id(self):
        byte_cmd = self.create_command(self.cmd_dev_id)
        await self.send_command(byte_cmd)
        [response_content, response_length, response_data_raw] = await self.read_response()
        # Parse response
        deviceDspId = response_data_raw[0:8].hex()
        aptinaHexId = response_data_raw[8:24].hex()
        aptinaId = ''
        # Reverse byte order of every 2 bytes
        for i in range(int(len(aptinaHexId)/4)):
            aptinaId += aptinaHexId[i*4+2] + aptinaHexId[i*4+3]
            aptinaId += aptinaHexId[i*4+0] + aptinaHexId[i*4+1]
        fw_data = list(struct.unpack('<24xH2x', response_data_raw))
        # Store information
        self.device_info.update({'deviceDspId':     deviceDspId,
                                 'aptinaId':        aptinaId,
                                 'device_id':       aptinaId[16:].upper(),
                                 'firmwareVersion': fw_data[0]})
        # 256-bit AES key
        # No idea whether this is right, but there is an indication in the Android app for it
        self.key = bytes.fromhex(aptinaId)
        return(self.device_info)
    
    # Scan data is returned in three batches # TODO
    async def scan(self):
        byte_cmd = self.create_command(self.cmd_scan)
        await self.send_command(byte_cmd)
        # Read scan
        rawdata = []
        for i in range(3):
            [response_content, response_length, response_data_raw] = await self.read_response()
            rawdata.append([response_content, response_length, response_data_raw])
        
        # Parse response (Store as base64 strings)
        self.scan_rawdata.update({'sample':          rawdata[1][2].hex(),
                                  'sample_dark':     rawdata[0][2].hex(), # The SCiO scans the dark reference first
                                  'sample_gradient': rawdata[2][2].hex()})
        self.scan_b64data.update({'sample':          self.encode_b64(rawdata[1][2]),
                                  'sample_dark':     self.encode_b64(rawdata[0][2]), # The SCiO scans the dark reference first
                                  'sample_gradient': self.encode_b64(rawdata[2][2])})
        return(self.scan_rawdata, self.scan_b64data)
    
    async def calibrate(self, calibration_fn = 'calibration.json'):
        temp_before      = await self.read_temperature()
        calibration_data = await self.scan()
        temp_after       = await self.read_temperature()
        self.calibration_data.update({'t_cmos_before':   temp_before['cmos_t'],
                                      't_chip_before':   temp_before['chip_t'],
                                      't_cmos_after':    temp_after['cmos_t'],
                                      't_chip_after':    temp_after['chip_t'],
                                      'sample':          calibration_data['sample'],
                                      'sample_dark':     calibration_data['sample_dark'],
                                      'sample_gradient': calibration_data['sample_gradient']})
        self.write_data_file(calibration_fn, calibration_data, temp_before, temp_after)
        return(self.calibration_data)
    
    def parse_scan(self):
        """Deprecated on purpose.

        This notebook captures and stores raw SCiO USB data only. Spectrum decoding experiments
        belong in notebooks/06_scio_offline_decode.ipynb, where they can be tested against
        data_scans/*.json without mixing analysis logic into the hardware-control notebook.
        """
        raise RuntimeError('Use notebooks/06_scio_offline_decode.ipynb for raw-to-spectrum decoding experiments.')
    
    def load_calibration_file(self, calibration_fn = 'calibration.json'):
        # Load the JSON file
        with open(calibration_fn, 'r') as file:
            json_data = json.load(file)

        # Load the "scan" object
        self.calibration_data = json_data['scan']
        
        return (self.calibration_data)
    
    def write_logfile(self, file_path, data, header):
        # Check if file already exists to determine whether to write header
        write_header = not os.path.isfile(file_path)
        
        # Open the file in write mode and create a CSV writer
        with open(file_path, mode='a', newline='') as f:
            writer = csv.writer(f)
            # Write header if file is new
            if write_header:
                writer.writerow(header)
            # Write data
            writer.writerow(data)
        return
    
    # Saves a single scan as raw data to JSON
    def write_data_file(self, output_fn, temp_before, temp_after):
        timestamp = datetime.datetime.utcnow().strftime('%Y-%m-%d %H:%M:%S')
        
        # Rename temperature reading dicts
        temp_before['t_cmos_before'] = temp_before.pop('cmos_t')
        temp_before['t_chip_before'] = temp_before.pop('chip_t')
        temp_before['t_obj_before']  = temp_before.pop('obj_t')
        temp_after['t_cmos_after']   = temp_after.pop('cmos_t')
        temp_after['t_chip_after']   = temp_after.pop('chip_t')
        temp_after['t_obj_after']    = temp_after.pop('obj_t')
        
        # Create the json string
        jsondata = {}
        jsondata['device'] = self.device_info
        # Add temperature
        jsondata['temperature'] = {}
        jsondata['temperature'].update(temp_before)
        jsondata['temperature'].update(temp_after)
        jsondata['temperature'].update({'sampled_at': timestamp})
        # Data, as hex and base64
        jsondata['raw_data'] = {'sampled_at': timestamp}
        jsondata['raw_data'].update(self.scan_rawdata)
        jsondata['b64_data'] = {'sampled_at': timestamp}
        jsondata['b64_data'].update(self.scan_b64data)
        
        
        # Write data to JSON file
        with open(output_fn, 'w') as outfile:
            json.dump(jsondata, outfile, indent=4)
        return
    
    def encode_b64(self, bytestring):
        urlSafeEncodedBytes = base64.urlsafe_b64encode(bytestring)
        urlSafeEncodedStr = str(urlSafeEncodedBytes, 'utf-8')
        return(urlSafeEncodedStr)
    
    async def run_scan(self, logfile, datafile, silent=True):
        if not silent: print('  - Connecting to device')
        await self.connect()
        
        if not silent: print('  - Reading BLE & device information')
        await self.read_ble_id()
        await self.read_device_id()
        for i in self.device_info:
            if(i == 'aptinaId'):
                if not silent: print('      ' + i + ':\t\t', self.device_info[i])
            else:
                if not silent: print('      ' + i + ':\t', self.device_info[i])
                pass
            pass
        
        if not silent: print('  - Reading battery status')
        battery_status = await self.read_battery()
        if not silent: print('      Battery charge:\t {:.0f}%'.format(battery_status['bat_charge_percent']))
        if not silent: print('      Battery is ' + battery_status['bat_charge_status'])
        
        # cmosTemperature, chipTemperature, objectTemperature
        if not silent: print('  - Reading temperature (before scan):')
        temp_before = await self.read_temperature()
        if not silent: print('      CMOS T:\t\t {:.3f}°C'.format(temp_before['cmos_t']))
        if not silent: print('      Chip T:\t\t {:.3f}°C'.format(temp_before['chip_t']))
        if not silent: print('      Obj. T:\t\t {:.3f}°C'.format(temp_before['obj_t']))
        
        if not silent: print('  - Scanning')
        raw_data, b64_data = await self.scan()
        print(b64_data)
        
        if not silent: print('  - Reading temperature (after scan):')
        temp_after = await self.read_temperature()
        if not silent: print('      CMOS T:\t\t {:.3f}°C'.format(temp_after['cmos_t']))
        if not silent: print('      Chip T:\t\t {:.3f}°C'.format(temp_after['chip_t']))
        if not silent: print('      Obj. T:\t\t {:.3f}°C'.format(temp_after['obj_t']))
        
        if not silent: print('  - Disconnecting')
        await self.disconnect()
        
        # TODO: save calibration data in the data file as well
        if not silent: print('  - Writing data to ' + datafile)
        self.write_data_file(datafile, temp_before, temp_after)
        
        #if not silent: print('  - Loading calibration file')
        #self.load_calibration_file()
        
        #if not silent: print('  - Parsing scan data')
        #reflectance = self.parse_scan()
        #print(reflectance)
        
        if not silent: print('  - Writing log to ' + logfile)
        timestamp_utc = datetime.datetime.utcnow()
        timestamp = datetime.datetime.strftime(timestamp_utc, '%Y-%m-%d %H:%M:%S')
        data = [str(timestamp)] +\
                list(self.device_info.values()) +\
               [battery_status['bat_charge_percent']] +\
                list(temp_before.values()) +\
                list(temp_after.values())
        header = ['timestamp'] +\
                  list(self.device_info.keys()) +\
                 ['battery_level',
                  't_cmos_before',
                  't_chip_before',
                  't_obj_before',
                  't_cmos_after',
                  't_chip_after',
                  't_obj_after']
        # TODO: Works, but I don't want output every time
        #self.write_logfile(logfile, data, header) # TODO: Works, but I don't want output every time
        
        return
    
    async def run_calibration(self, silent=True):
        if not silent: print('  - Connecting to device')
        await self.connect()
        
        if not silent: print('  - Reading BLE & device information')
        await self.read_ble_id()
        await self.read_device_id()
        for i in self.device_info:
            if(i == 'aptinaId'):
                if not silent: print('      ' + i + ':\t\t', self.device_info[i])
            else:
                if not silent: print('      ' + i + ':\t', self.device_info[i])
                pass
            pass
        
        if not silent: print('  - Reading battery status')
        battery_status = await self.read_battery()
        if not silent: print('      Battery charge:\t {:.0f}%'.format(battery_status['bat_charge_percent']))
        if not silent: print('      Battery is ' + battery_status['bat_charge_status'])
        
        if not silent: print('  - Running calibration scan')
        # TODO: Insert the ready-for-WR command here?
        #await self.calibrate()
        
        if not silent: print('  - Disconnecting')
        await self.disconnect()
        
        return
    
    async def run_test(self, silent=True):
        if not silent: print('  - Connecting to device')
        await self.connect()
        
        if not silent: print('  - Testing other commands:')
        # Options to test:
        raw_resp = await self.test_cmd(self.cmd_dev_status) # Responds with 8 bytes, starting with 05 and followed by 0s
        
        print(raw_resp)
        print('Hex:', raw_resp[2].hex())
        
        if not silent: print('  - Disconnecting')
        await self.disconnect()
        
        print('    Int:', struct.unpack('<I', raw_resp[2]), '    Shorts:', struct.unpack('<2H', raw_resp[2]))
        
        #out = await self.test_parse_response(raw_resp)
        #print(out)
        return
    
    # Generic commands, can be used for testing
    async def test_cmd(self, cmd):
        byte_cmd = self.create_command(cmd)
        await self.send_command(byte_cmd)
        [response_content, response_length, response_data_raw] = await self.read_response()
        return([response_content, response_length, response_data_raw])
    
    # Generic, for testing
    async def test_parse_response(self, response_data_raw):
        # Parse response
        print('As hex:', response_data_raw[2].hex())
        print(len(response_data_raw[2]))
        
        try:
            data_struct = '<4I' # File list
            resp = list(struct.unpack(data_struct, response_data_raw[2]))
            print(resp)
            file_list = {}
            for file_nb in range(0, int(len(resp)/2)):
                if((resp[file_nb*2] >= 87) & (resp[file_nb*2] <= 95)):
                    file_list.update({file_nb:{'file_type':resp[file_nb*2], 'file_version':resp[file_nb*2+1]}})
                    key_start_byte = file_nb*8
                    key_end_byte   = file_nb*8+4
                    val_start_byte = file_nb*8+4
                    val_end_byte   = file_nb*8+8
                    attr = {'key':response_data_raw[2][key_start_byte:key_end_byte].hex(),
                            'value':response_data_raw[2][val_start_byte:val_end_byte].hex()}
                    file_list.update({file_nb:attr})
                else:
                    attr = {'file_type':resp[file_nb*2], 'file_version':resp[file_nb*2+1]}
                    file_list.update({file_nb:attr})
                pass
            #print(file_list)
            
            #file_list.update({'file':1, 'file_type':resp[2], 'file_version':resp[3]})
            
            
        except:
            await self.disconnect()
            raise
        #dev_id = struct.unpack(data_struct, response_data_raw)
        return(resp)
```
